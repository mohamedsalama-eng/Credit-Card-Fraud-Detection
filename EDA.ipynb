```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

# 1) Normalized overlay: compares the *shape* of the distributions
for label, name, color in [(0, "Normal", "tab:blue"), (1, "Fraud", "tab:red")]:
    axes[0].hist(df.loc[df["Class"] == label, "Time"], bins=48,
                 density=True, alpha=0.5, label=name, color=color)
axes[0].set_xlabel("Time (seconds)")
axes[0].set_ylabel("Density")
axes[0].set_title("Time distribution by Class (normalized)")
axes[0].legend()

# 2) Fraud-only raw counts
axes[1].hist(df.loc[df["Class"] == 1, "Time"], bins=48, color="tab:red", alpha=0.8)
axes[1].set_xlabel("Time (seconds)")
axes[1].set_ylabel("Fraud count")
axes[1].set_title("Fraud counts over time")

plt.tight_layout()
plt.show()
```

```python
df["Hour"] = ((df["Time"] // 3600) % 24).astype(int)

fig, axes = plt.subplots(1, 2, figsize=(14, 4))

# Histogram: Time in hours by Class (normalized, so fraud is visible)
for label, name, color in [(0, "Normal", "tab:blue"), (1, "Fraud", "tab:red")]:
    axes[0].hist(df.loc[df["Class"] == label, "Hour"], bins=24, range=(0, 24),
                 density=True, alpha=0.5, label=name, color=color)
axes[0].set_xlabel("Hour of day")
axes[0].set_ylabel("Density")
axes[0].set_title("Hour distribution by Class")
axes[0].set_xticks(range(0, 24, 2))
axes[0].legend()

# Bar plot: fraud rate per hour
fraud_rate = df.groupby("Hour")["Class"].mean() * 100
axes[1].bar(fraud_rate.index, fraud_rate.values, color="tab:red", alpha=0.8)
axes[1].set_xlabel("Hour of day")
axes[1].set_ylabel("Fraud rate (%)")
axes[1].set_title("Fraud rate by hour")
axes[1].set_xticks(range(0, 24, 2))

plt.tight_layout()
plt.show()
```

```python
counts = df.groupby(["Hour", "Class"]).size().unstack(fill_value=0)
counts.plot(kind="bar", figsize=(10, 4), logy=True, title="Transactions per hour by Class")
plt.xlabel("Hour of day")
plt.ylabel("Count (log scale)")
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(24, 8))

# 1) Amount distribution (raw): heavily right-skewed
axes[0].hist(df["Amount"], bins=100, color="tab:blue", alpha=0.8)
axes[0].set_yscale("log")
axes[0].set_xlabel("Amount")
axes[0].set_ylabel("Count (log scale)")
axes[0].set_title("Amount distribution")

# 2) Amount distribution after log transform: reveals the real shape
axes[1].hist(np.log1p(df["Amount"]), bins=60, color="tab:green", alpha=0.8)
axes[1].set_xlabel("log(1 + Amount)")
axes[1].set_ylabel("Count")
axes[1].set_title("Log-transformed Amount")

# 3) Amount by Class (normalized, so fraud is visible)
for label, name, color in [(0, "Normal", "tab:blue"), (1, "Fraud", "tab:red")]:
    axes[2].hist(np.log1p(df.loc[df["Class"] == label, "Amount"]), bins=60,
                 density=True, alpha=0.5, label=name, color=color)
axes[2].set_xlabel("log(1 + Amount)")
axes[2].set_ylabel("Density")
axes[2].set_title("Amount by Class (log scale, normalized)")
axes[2].legend()

plt.tight_layout()
plt.show()
```

```python
df["Amount_log"] = np.log1p(df["Amount"])

fig, axes = plt.subplots(1, 3, figsize=(16, 4))

axes[0].boxplot(df["Time"], vert=False, flierprops=dict(marker=".", markersize=2, alpha=0.3))
axes[0].set_title("Time (seconds)")
axes[0].set_xlabel("Seconds")

axes[1].boxplot(df["Amount"], vert=False, flierprops=dict(marker=".", markersize=2, alpha=0.3))
axes[1].set_title("Amount")
axes[1].set_xlabel("Amount")

axes[2].boxplot(df["Amount_log"], vert=False, flierprops=dict(marker=".", markersize=2, alpha=0.3))
axes[2].set_title("Log Amount (log1p)")
axes[2].set_xlabel("log(1 + Amount)")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

feats = df.columns.drop("Class")

n_cols = 5
n_rows = int(np.ceil(len(feats) / n_cols))
fig, axes = plt.subplots(n_rows, n_cols, figsize=(20, 3 * n_rows))
axes = axes.ravel()

for ax, c in zip(axes, feats):
    ax.hist(df[c], bins=50, color="tab:blue", alpha=0.8)
    ax.set_title(c, fontsize=9)

for ax in axes[len(feats):]:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
skew = df[feats].skew().round(3)

skew_table = pd.DataFrame({
    "skew": skew,
    "abs_skew": skew.abs(),
    "kurtosis": df[feats].kurt().round(3),
}).sort_values("abs_skew", ascending=False).drop(columns="abs_skew")

print(skew_table.to_string())
```

```python
corr = df.corr()

plt.figure(figsize=(16, 12))
sns.heatmap(corr, cmap="coolwarm", center=0, vmin=-1, vmax=1, square=True,
            linewidths=0.3, cbar_kws={"shrink": 0.7})
plt.title("Correlation heatmap (train)")
plt.tight_layout()
plt.show()
```
